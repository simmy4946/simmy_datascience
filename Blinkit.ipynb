**outliers treatment**

```python
#Outlier Treatment: Removing top 5% & bottom 1% of the Column Outlier values
Q3 = df['distance'].quantile(0.75)
Q1 = df['distance'].quantile(0.25)
IQR = Q3-Q1
df = df[(df['distance'] > Q1 - 1.5*IQR) & (df['distance'] < Q3 + 1.5*IQR)]

sns.boxplot(data=df, x="distance", orient='h')
plt.show()
#Outlier Handling for Miles Feature
#Outlier Treatment: Removing top 5% & bottom 1% of the Column Outlier values
Q3 = df['duration'].quantile(0.75)
Q1 = df['duration'].quantile(0.25)
IQR = Q3-Q1
df= df[(df['duration'] > Q1 - 1.5*IQR) & (df['duration'] < Q3 + 1.5*IQR)]

sns.boxplot(data=df, x="duration", orient='h')
plt.show()
```

Observation:

There are many outliers we have treated them with outliers handling methods with the removal of top of 5% and bottom 1% of distance values are removed by the treatment of filtering distances that was presented outside 1.5 times the IQR from the 1st and the 3rd quartiles.

similarly, in the duartion outliers we treated them in a same way with top 5% and bottom 1% of duration values are removed with same IQR method.

To calculate the percentage deviation between the actual metered_price and the upfront_price for each ride and identify where significant deviations (> 20%) occur..

Formula:
Price Deviation (%)
=
(
Actual Price
−

Upfront Price
)

100
Price Deviation (%)=(
Upfront Price
Actual Price−Upfront Price
​
 )×100 **

```python
# Calculate the percentage deviation between metered_price (actual) and upfront_price
df['price_deviation_pct'] = ((df['metered_price'] - df['upfront_price']) / df['upfront_price']) * 100

# Filter for rides where the percentage deviation is greater than 20%
significant_deviation = df[df['price_deviation_pct'].abs() > 20]
```

```python
significant_deviation
```

```python
# mean of significant deviation:
significant_deviation['price_deviation_pct'].mean()
```

```python
# median of significant deviation:
significant_deviation['price_deviation_pct'].median()
```

If the median value is positive, it shows that the majority of these significant deviations show that the actual price is higher than the upfront price.

If the median value is negative, it shows that the majority of these significant deviations show that the upfront price is higher than the actual price.

Imply:
This significant median deviation indicates that the upfront prices are generally much lower than the actual prices. This indicates that the upfront pricing model is underestimating the cost of rides.

 There's a gap between the upfront and actual prices, pointing to a need for improving the accuracy of the pricing model or adjusting the upfront pricing strategy.

 conclusion: For rides with large deviations, the actual price tends to be about 64% higher than the upfront price. This shows that the discrepancy and give a opinion for a review and adjustment of the upfront pricing mechanism may be necessary.
